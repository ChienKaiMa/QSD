```python
def plot_psucc_noise_tol(tol, noise_level, psucc, tag, num_levels):
    # Prevent segmentation fault
    # plt.switch_backend("Tkagg")
    fig = plt.figure(dpi=1200)
    fig.set_size_inches(3.2, 2.7)

    # Contour plot
    x = tol
    y = noise_level
    X, Y = np.meshgrid(x, y)
    Z = psucc

    # contour or contourf
    # TODO
    # Different levels
    # contour = plt.contour(X, Y, Z, levels=15, cmap="viridis")

    # https://stackoverflow.com/questions/53641644/set-colorbar-range-with-contourf
    vmin = -0
    vmax = 1
    levels = np.linspace(vmin, vmax, num_levels + 1)

    # https://matplotlib.org/stable/users/explain/colors/colormaps.html
    contour = plt.contour(
        X,
        Y,
        Z,
        levels=levels,
        # cmap="viridis",
        # cmap="cool",
        cmap="seismic",
        vmin=vmin,
        # vmax=vmax,
        extend="both",
        edgecolor="white",
        alpha=0.8,
    )

    # https://stackoverflow.com/questions/73886009/contourf-plots-with-streamline-numbers
    plt.clabel(contour, inline=True, fontsize=4, fmt="%.3f", colors="black")

    # Add a color bar to show the scale of Z
    plt.colorbar(
        contour,
        ticks=linspace(vmin, vmax, 11),
        label="Success probability",
    )

    # Add labels and title
    plt.xlabel(r"Tolerance $\alpha, \beta$")
    plt.ylabel("Depolarizing noise level")

    # TODO
    plt.title(f"{tag}")

    plt.grid(True)
    plt.tight_layout()

    # Save the plot
    plt.savefig(f"../results/psucc_noise_tol_{tag}.png")
    plt.close()

    return
```

## Check the boundary values

Since the solver precision is not perfect, we try to see how we can prevent results that are not reasonable.
For example, we change the constraint of 0 to a small positive number like 1e-8.

```python
np.set_printoptions(threshold=np.inf)
print(np.where(pinc_q2 > 1.0000001))
print(pinc_q2[np.where(pinc_q2 > 1.0000001)])
print(pinc_q2[np.where(pinc_q2 > 1.00000001)])
# print(pinc_q2[np.where(pinc_q2 > 1.000000001)])
```

We can see that a lot more bad results (total prob > 1 + epsilon) emerge when epsilon decreases to 1e-8.

```python
print(np.where(pinc_q2 < 0))
print(pinc_q2[np.where(pinc_q2 < 0)])
```

## Trends in 3D

```python
def plot_3d_prob(points, tag, noise_level):
    """Verify that the total probability of the POVM sums to identity."""
    fig = plt.figure(dpi=600)
    # fig.set_size_inches(3.2, 2.7)
    fig.set_size_inches(7, 6)

    ax = fig.add_subplot(111, projection="3d")

    for point in points:
        ax.scatter(point[0], point[1], point[2])

    ax.set_zlim(-0.01, 1.01)

    # Labeling axes
    ax.set_xlabel("Depolarizing noise level")
    ax.set_ylabel(r"Tolerance $\alpha$, $\beta$")
    # ax.set_zlabel("Inconclusive probability")
    ax.set_zlabel("Total success probability")

    # ax.legend(f"Noise level {noise_level}")

    # Adjust the view angle
    # ax.view_init(elev=30.0, azim=30)
    ax.view_init(elev=30.0)
    # plt.show()
    plt.tight_layout()
    plt.savefig(f"../results/3d_prob_{tag}.png")
    plt.close()
    return
```

```python
def plot_surf(X, Y, Z, tag):
    fig = plt.figure(dpi=200)
    # fig.set_size_inches(3.2, 2.7)
    fig.set_size_inches(7, 6)

    ## Make data
    Z.reshape(X.shape[0], Y.shape[0])
    X, Y = np.meshgrid(X, Y)
    ax = fig.add_subplot(111, projection="3d")

    # Plot the surface with heatmap-like coloring
    # https://stackoverflow.com/questions/51574861/plotting-3d-surface-using-python-raise-valueerrorargument-z-must-be-2-dimensi
    # https://stackoverflow.com/questions/55577981/plotly-plot-trisurf-isnt-working-with-arange-arrays

    surf = ax.plot_trisurf(
        X.flatten(),
        Y.flatten(),
        Z.flatten(),
        cmap="viridis",
        edgecolor="none",
        vmin=0,
        vmax=1,
    )  # Remove edges to give a smoother look

    # Customize the z axis
    ax.set_zlim(-0.01, 1.01)

    # Add a color bar which maps values to colors
    cbar = fig.colorbar(surf, shrink=0.5, aspect=5)
    cbar.set_label("Z Value")  # Label for the colorbar

    # Labeling axes
    ax.set_xlabel("Depolarizing noise level")
    ax.set_ylabel(r"Tolerance $\alpha$, $\beta$")
    ax.set_zlabel("Total success probability")

    # Adjust the view angle
    ax.view_init(elev=30.0, azim=-120)
    # ax.view_init(elev=0, azim=-90, roll=0)
    plt.tight_layout()
    plt.show()
    # plt.savefig("3dplot_color.png")

    plt.close()

    return
```
